```python
# Step 3.1: Handling Missing Values

# Create a copy of the original dataset for preprocessing
df_missing = df.copy()

# Introduce a few missing values for demonstration
df_missing.loc[3, "Customer_Rating"] = np.nan
df_missing.loc[8, "Discount_Percent"] = np.nan
df_missing.loc[15, "Customer_Rating"] = np.nan

# Display the dataset before handling missing values
print("Dataset before handling missing values:")
display(df_missing.head(10))

# Detect and report missing values
print("\nMissing values in each column:")
print(df_missing.isnull().sum())
```

```python
# Fill missing numerical values using the column mean

df_missing["Customer_Rating"] = df_missing["Customer_Rating"].fillna(
    df_missing["Customer_Rating"].mean()
)

df_missing["Discount_Percent"] = df_missing["Discount_Percent"].fillna(
    df_missing["Discount_Percent"].mean()
)

# Check missing values after treatment
print("Missing values after handling:")
print(df_missing.isnull().sum())

# Display the affected rows after handling missing values
print("\nDataset after handling missing values:")
display(df_missing.loc[[3, 8, 15]])
```

```python
# Step 3.2: Outlier Detection Using IQR

# Select numerical columns for outlier analysis
numeric_columns = [
    "Quantity",
    "Unit_Price",
    "Discount_Percent",
    "Customer_Rating",
    "Total_Sales"
]

# Calculate Q1, Q3, and IQR
Q1 = df[numeric_columns].quantile(0.25)
Q3 = df[numeric_columns].quantile(0.75)
IQR = Q3 - Q1

# Calculate lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Display the IQR calculations
iqr_table = pd.DataFrame({
    "Q1": Q1,
    "Q3": Q3,
    "IQR": IQR,
    "Lower_Bound": lower_bound,
    "Upper_Bound": upper_bound
})

print("IQR Outlier Analysis:")
display(iqr_table)
```

```python
# Identify outliers using the IQR boundaries

outlier_mask = (
    (df[numeric_columns] < lower_bound) |
    (df[numeric_columns] > upper_bound)
)

print("Number of outliers by column:")
print(outlier_mask.sum())

print("\nRows containing at least one outlier:")
display(df[outlier_mask.any(axis=1)])
```

```python
# Remove rows containing IQR outliers

df_no_outliers = df[~outlier_mask.any(axis=1)].copy()

print("Original dataset shape:", df.shape)
print("Dataset shape after outlier removal:", df_no_outliers.shape)

# Display the cleaned dataset
display(df_no_outliers.head())
```

```python
# Step 3.3: Data Reduction

# Use the outlier-cleaned dataset as the starting point
df_reduced = df_no_outliers.copy()

print("Dataset before data reduction:")
print("Rows:", df_reduced.shape[0])
print("Columns:", df_reduced.shape[1])

display(df_reduced.head())
```

```python
# Randomly sample 80% of the rows
df_reduced = df_reduced.sample(frac=0.80, random_state=42)

# Drop the Product_ID column because it is an identifier
df_reduced = df_reduced.drop(columns=["Product_ID"])

# Reset the index
df_reduced = df_reduced.reset_index(drop=True)

print("Dataset after data reduction:")
print("Rows:", df_reduced.shape[0])
print("Columns:", df_reduced.shape[1])

display(df_reduced.head())
```

```python
# Step 3.4: Data Scaling and Discretization

print("Dataset before scaling and discretization:")
display(df_reduced[[
    "Quantity",
    "Unit_Price",
    "Customer_Rating"
]].head(10))
```

```python
# Apply Min-Max Scaling to Unit_Price

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df_reduced["Unit_Price_Scaled"] = scaler.fit_transform(
    df_reduced[["Unit_Price"]]
)

# Discretize Customer_Rating into meaningful categories
df_reduced["Rating_Category"] = pd.cut(
    df_reduced["Customer_Rating"],
    bins=[0, 3.9, 4.4, 5.0],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)

# Display the results
print("Dataset after scaling and discretization:")

display(df_reduced[[
    "Quantity",
    "Unit_Price",
    "Unit_Price_Scaled",
    "Customer_Rating",
    "Rating_Category"
]].head(10))
```

```python
# Step 4.2: Central Tendency Measures

# Select numerical columns
numeric_df = df.select_dtypes(include=np.number)

# Calculate central tendency measures
central_tendency = pd.DataFrame({
    "Minimum": numeric_df.min(),
    "Maximum": numeric_df.max(),
    "Mean": numeric_df.mean(),
    "Median": numeric_df.median(),
    "Mode": numeric_df.mode().iloc[0]
})

print("CENTRAL TENDENCY MEASURES")
print("=" * 70)

display(central_tendency.round(2))
```

### Interpretation of Central Tendency

The central tendency measures summarize the typical values and overall position of the numerical variables in the retail sales dataset. The mean represents the average value, while the median represents the middle observation after the values are ordered. The minimum and maximum values show the lower and upper limits of each numerical variable, while the mode identifies the most frequently occurring value. Together, these measures provide a basic understanding of the typical characteristics of the dataset.

```python
# Step 4.3: Dispersion Measures

# Calculate dispersion measures
dispersion = pd.DataFrame({
    "Minimum": numeric_df.min(),
    "Q1 (25%)": numeric_df.quantile(0.25),
    "Q2 (50%)": numeric_df.quantile(0.50),
    "Q3 (75%)": numeric_df.quantile(0.75),
    "Maximum": numeric_df.max(),
    "Range": numeric_df.max() - numeric_df.min(),
    "IQR": numeric_df.quantile(0.75) - numeric_df.quantile(0.25),
    "Variance": numeric_df.var(),
    "Standard_Deviation": numeric_df.std()
})

print("DISPERSION MEASURES")
print("=" * 100)

display(dispersion.round(2))
```

### Interpretation of Dispersion Measures

The dispersion measures describe how widely the numerical observations are distributed. The range represents the difference between the maximum and minimum values, while the quartiles divide the data into four sections. The interquartile range measures the spread of the middle 50 percent of observations and is also useful for identifying potential outliers. Variance and standard deviation measure the overall variability of the numerical variables, with larger values indicating greater spread around the mean.

```python
# Step 4.4: Correlation Analysis

# Calculate the correlation matrix for numerical variables
correlation_matrix = numeric_df.corr()

print("CORRELATION MATRIX")
print("=" * 70)

display(correlation_matrix.round(2))
```

### Interpretation of Correlation Analysis

The correlation matrix measures the strength and direction of linear relationships between the numerical variables. Correlation values close to +1 indicate a strong positive relationship, values close to -1 indicate a strong negative relationship, and values close to 0 indicate a weak or limited linear relationship. The matrix allows relationships between quantity, unit price, discount percentage, customer rating, and total sales to be examined. These results provide an overview of how the numerical variables move in relation to one another.
